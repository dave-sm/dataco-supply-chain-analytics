## Power BI Deliverables

Six dashboards map to the SCOR phases and the eight analytical notebooks. Each dashboard answers a specific operational question for a defined audience.

The interactive `SmartSupplyChainSCOR.pbix` file is available upon request and is not included in the repo. All metric definitions are available in `power-bi/dax_measures.txt`. Model assumptions and relationships are documentd in `power-bi/model_notes.md`.

### Dashboards

#### 1. Strategic Overview (SCOR Scorecard)

**Purpose:** Executive-level supply chain health monitoring. Single-page view of the metrics that matter for monthly and quarterly operational reviews. Answers: *Is the network generating value and retaining it?*   
**Audience:** Executives.   
**Source notebooks:** 01, 02, 04.     



---

#### 2. Demand and Forecasting (Plan / Deliver)

**Purpose:** Forward-looking capacity planning and inventory positioning. Supports the transition from descriptive tracking to predictive planning. Answers: *Where is demand heading, and do ML and baseline forecasts forecasts agree?*   
**Audience:** Operations.   
**Source notebook:** 03.   



---

#### 3. Financial Performance and ROI (Measure / Return)

**Purpose:** Profitability diagnostics and margin recovery scenario modelling. Identifies where profit is created, diluted, or destroyed. Answers: *Where is profit created, and where is it leaking?*   
**Audience:** Executives + Operations.      
**Source notebook:** 04.    



---

#### 4. Operational Risk (Source / Make / Deliver)

**Purpose:** Proactive late-delivery risk management. Converts risk scores into prioritised intervention triggers. Answers: *Which shipments warrant intervention, and what drives their risk?*   
**Audience:** Operations.      
**Source notebook:** 05.    



---

#### 5. Global Logistics and Bottlenecks (Deliver)

**Purpose:** Bottleneck identification and SLA calibration decision support. The central dashboard for the "Promise vs. Physics" finding. Answers: *Where is the network over-promising, and what should be fixed first?*   
**Audience:** Operations.      
**Source notebook:** 06.   



---

#### 6. Customer Service Tiers (Plan / Deliver / Return)

**Purpose:** Customer-centric service allocation and retention monitoring. Aligns operational effort with customer value and risk exposure.  Answers: *Are high-value customers treated differently, and where is the service-design correction needed?*   
**Audience:** Executives + Analyst.      
**Source notebook:** 07.
